```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Loading dataset from github
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user"
df = pd.read_csv(url)
df = pd.read_csv(url, sep='|')
```

```python
#Renaming columns for easy access
df.columns = ['user_id', 'age', 'gender', 'occupation', 'zip_code']
```

```python
#basic details about the dataset
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
#Data preprocessing
#Remove Duplicates
df = df.drop_duplicates()
```

```python
#DATA VISUALIZATION

# Age distribution
plt.figure()
sns.histplot(df['age'], bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
# Gender distribution
plt.figure()
sns.countplot(x='gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
# Top occupations
plt.figure()
df['occupation'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Occupations")
plt.show()
```

```python
# Age vs Gender
plt.figure()
sns.boxplot(x='gender', y='age', data=df)
plt.title("Age vs Gender")
plt.show()
```

```python
print("""
KEY INSIGHTS:
1. Most users are between 20–40 years old.
2. Male users are more than female users.
3. Students, engineers, and educators are top occupations.
4. Age distribution is slightly right-skewed.
5. Occupation distribution is highly imbalanced.
""")
```

```python
#Save the cleaned data
df.to_csv("cleaned_user_data.csv", index=False)
```
